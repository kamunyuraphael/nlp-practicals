10. Practical Example: Building Bigrams
Consider this sentence:
***Students use computers for learning
The bigrams are:
Students use
use computers
computers for
for learning


```python
# Manual bigram construction (matches the worked example above)
sentence = "Students use computers for learning"
words = sentence.split()

bigrams = [(words[i], words[i + 1]) for i in range(len(words) - 1)]

for w1, w2 in bigrams:
    print(w1, w2)
```

### Generalising to any n
Rather than writing the bigram loop by hand each time, NLTK's `ngrams` utility
builds unigrams, bigrams, trigrams, etc. from any tokenised text.

```python
from nltk.tokenize import word_tokenize
from nltk.util import ngrams


def generate_ngrams(text, n):
    tokens = word_tokenize(text)
    return list(ngrams(tokens, n))


text = "Students use computers for learning"

print("Unigrams:", generate_ngrams(text, 1))
print("Bigrams:", generate_ngrams(text, 2))
print("Trigrams:", generate_ngrams(text, 3))
```

### Why n-grams matter
Bigrams and trigrams capture local word order, which is the basis of simple
language models (predicting the next word), collocation detection (which word
pairs occur together more than chance would suggest), and features for text
classification.

## Week 2 Tool Activity
Accepts a paragraph, lowercases it, tokenises it, builds unigrams/bigrams/trigrams, counts their frequencies, and displays the five most frequent n-grams of each type.

```python
"""
Week 2 Tool Activity
N-gram Frequency Analyzer

Accepts a paragraph, lowercases it, tokenises it, builds unigrams/bigrams/
trigrams, counts their frequencies, and displays the five most frequent
n-grams of each type.
"""

from collections import Counter

from nltk.tokenize import word_tokenize
from nltk.util import ngrams


def analyze_paragraph(paragraph, top_n=5):
    # 1. Accept a paragraph (function argument)
    # 2. Convert to lowercase
    lowered = paragraph.lower()

    # 3. Tokenise
    tokens = word_tokenize(lowered)
    tokens = [t for t in tokens if t.isalpha()]  # drop punctuation tokens

    # 4-6. Generate unigrams, bigrams, trigrams
    unigrams = list(ngrams(tokens, 1))
    bigrams = list(ngrams(tokens, 2))
    trigrams = list(ngrams(tokens, 3))

    # 7. Count frequencies
    unigram_freq = Counter(unigrams)
    bigram_freq = Counter(bigrams)
    trigram_freq = Counter(trigrams)

    # 8. Five most frequent of each
    results = {
        "Unigrams": unigram_freq.most_common(top_n),
        "Bigrams": bigram_freq.most_common(top_n),
        "Trigrams": trigram_freq.most_common(top_n),
    }
    return results, tokens


def print_table(title, ngram_counts):
    print(f"\nTop {len(ngram_counts)} {title}")
    print(f"| {'N-gram':<20}| {'Frequency':<10}|")
    print(f"|{'-'*21}|{'-'*11}|")
    for gram, freq in ngram_counts:
        phrase = " ".join(gram)
        print(f"| {phrase:<20}| {freq:<10}|")


if __name__ == "__main__":
    paragraph = (
        "Students use computers every day. Students use computers for "
        "learning, and students use computers for research too. Computer "
        "science students use computers to write programs and build "
        "software. Many students learn computer science because computer "
        "science helps students understand modern technology. Students "
        "learn to use computers effectively, and computer science students "
        "learn algorithms quickly."
    )

    results, tokens = analyze_paragraph(paragraph)

    print("Tokens:", tokens)
    for title, counts in results.items():
        print_table(title, counts)
```

```python
paragraph = (
    "Students use computers every day. Students use computers for "
    "learning, and students use computers for research too. Computer "
    "science students use computers to write programs and build "
    "software. Many students learn computer science because computer "
    "science helps students understand modern technology. Students "
    "learn to use computers effectively, and computer science students "
    "learn algorithms quickly."
)

results, tokens = analyze_paragraph(paragraph)

print("Tokens:", tokens)
for title, counts in results.items():
    print_table(title, counts)
```

### Notes
- Punctuation tokens are dropped before building n-grams (`t.isalpha()`), so counts reflect words only.
- "Five most frequent N-grams" is shown **per n-gram type** (unigram, bigram, trigram) rather than pooled together, since mixing n-gram sizes in one ranking isn't a fair comparison — a bigram will almost always be rarer than a unigram.
